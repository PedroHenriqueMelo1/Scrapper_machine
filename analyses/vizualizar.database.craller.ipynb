```python
import pandas as pd 
from sqlalchemy import create_engine


caminho_db = "../imoveis.db"

# 2. Crie a engine de conexão do SQLAlchemy
# O formato da string de conexão para SQLite é 'sqlite:///caminho/do/arquivo.db'
engine = create_engine(f"sqlite:///{caminho_db}")

# 3. Leia uma tabela inteira para um DataFrame do Pandas
query = "SELECT * FROM imoveis"
df = pd.read_sql(query, con=engine)

# Exibe as primeiras linhas do DataFrame
print(df.head())

pd.set_option("display.max_columns", None)  # Mostra todas as colunas
pd.set_option("display.width", 1000)  # Alarga a largura total da consola
pd.set_option(
    "display.max_colwidth", None
)  # 

pd.read_sql_table("imoveis", con=engine)

print(df.head())
```
